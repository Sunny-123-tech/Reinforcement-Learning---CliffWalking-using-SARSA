## SARSA

```python
alpha = 0.5  # See vd SARSA implementation
gamma = 0.99 
epsilon = 0.1 # 10% exploration 90% exploitation
episodes = 500  # round from st pt to ending pt
```

```python
# Q-table => An grid which stores Q values
Q = np.zeros((48, 4)) 
```

```python
# Policy - epsilon greedy : state => action it simply means what action should be taken based on state 

def epsilon_greedy(state):
    if random.random() < epsilon:     # random is generating a random no.
        return env.action_space.sample()  # Exploration part and action means as it is exploration then it can go anywhere up,down,left,right
    else:
        return np.argmax(Q[state])    # returns max index of state Q's val
```

```python
# without visual env

for episode in range(episodes):
    render = (episode % 50 == 0)
    
    if render:
        env = gym.make("CliffWalking-v1", render_mode = "human")
        print(f"rendering env for episodes = {episode}")
    else:
        env = gym.make("CliffWalking-v1")
        
    done = False  # when the episode will end means either terminal or truncate
    state, _ = env.reset()
    action = epsilon_greedy(state)  # on policy
    
    total_reward = 0
    episode_length = 0

    while not done:
        next_state, reward, terminated, truncated, _ = env.step(action)  # Updation
        done = terminated or truncated
        next_action = epsilon_greedy(next_state)

        # SARSA Update
        Q[state , action] += alpha * (reward + gamma* Q[next_state, next_action] - Q[state, action])

        state = next_state
        action = next_action
        
        total_reward += reward
        episode_length += 1

    print(f"For episode = {episode}/500, tot_reward = {total_reward} & Episode_len = {episode_length}")
    env.close()
```

```python
# What our agent learn after 500 episodes (With visualization)

env = gym.make("CliffWalking-v1", render_mode = "human")
state, _ = env.reset()
done = False
total_reward = 0
epsilon_len = 0

while not done:
    action = np.argmax(Q[state])
    next_state, reward, terminated, truncated, _ = env.step(action)
    done = terminated or truncated

    state = next_state
    
    total_reward += reward
    episode_length += 1 

print(f"total_reward = {total_reward} & episode_len = {episode_length}")
env.close()
```
